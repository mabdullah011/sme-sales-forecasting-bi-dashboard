# 02 - Exploratory Data Analysis
Analyze monthly trend, category performance, regional sales, top products, and customer behavior.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.feature_engineering import aggregate_monthly_sales

df = pd.read_csv('../data/processed/retail_sales_cleaned.csv', parse_dates=['order_date'])
monthly = aggregate_monthly_sales(df)
sns.lineplot(data=monthly, x='order_date', y='monthly_sales')
plt.title('Monthly Sales Trend')
plt.xticks(rotation=45)
plt.tight_layout()
```
